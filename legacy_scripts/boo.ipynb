```python
path = "/content/drive/MyDrive/BMG/corridas/4754755/cooling/10/boo/"
```

```python
z = np.loadtxt(path+'dump.110000000.cc')
zr = 1
cu = 2
ag = 3
ztypes = z[:,1]
q4z = []
q6z = []
q4c = []
q6c = []
q4a = []
q6a = []
```

```python
for i, ztype  in enumerate(ztypes):
    if ztype == zr:
        q4z.append(z[i,2])
        q6z.append(z[i,3])
    if ztype == cu:
        q4c.append(z[i,2])
        q6c.append(z[i,3])
    if ztype == ag:
        q4a.append(z[i,2])
        q6a.append(z[i,3])
```

```python
plt.plot(q4z,q6z,'b*')
plt.plot(q4c,q6c,'g*')
plt.plot(q4a,q6a,'r*')
plt.show()
```

```python
num_bins = 100
n, bins, patches = plt.hist(q6z, num_bins, facecolor ='blue')
n1, bins1, patches1 = plt.hist(q6c, num_bins, facecolor ='green')
n2, bins2, patches2 = plt.hist(q6a, num_bins, facecolor ='red')
plt.plot(bins,bins1)
plt.show()
```
