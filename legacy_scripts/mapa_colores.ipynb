```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
path = "/content/drive/MyDrive/BMG/corridas/4754755/cooling/10/gofr/"
```

```python
end_dat = 1710
temp_ini = 2000
temp_fin = 300
temp = np.linspace(temp_ini,temp_fin,end_dat)
g = []
temp = temp.astype(int)
for i in range(end_dat):
   z = np.loadtxt(path + 'gofr4754755.data'+ str(i)+'_gofr.dat')
   g.append(z[:,4])
```

```python
X, Y =  np.meshgrid(z[:,1],temp)
l1 = plt.MaxNLocator(nbins=20).tick_values(0,3)
l2 = plt.MaxNLocator(nbins=55).tick_values(0,3)
fig, (ax0, ax1) =plt.subplots(nrows = 2)
im  = ax0.contourf(X, Y, g, levels = l1, cmap = 'plasma')
ax0.set_xlim(1.5,6.3)
cs = ax1.contourf(X, Y, g, levels = l2, cmap = 'plasma')
fig.tight_layout()
ax1.set_xlim(1.5,6.3)
ax1.set_xlabel(r'g(r)')
ax1.set_ylabel(r'$T\mathrm{(K)}$')
plt.show()
```

```python
fig2, ax2 =plt.subplots()
fig2.tight_layout()
cs = ax2.contourf(X, Y, g ,l1, cmap = 'OrRd')
#cbar = ax2.colorbar()
fig2.colorbar(cs)
ax2.set_xlabel(r'g(r)')
ax2.set_ylabel(r'$T\mathrm{(K)}$')
ax2.set_xlim(1.5,6.3)
plt.show()
```
