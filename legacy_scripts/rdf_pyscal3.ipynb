```python
import pyscal3 as pc
import matplotlib.pyplot as plt
```

```python
traj = pc.Trajectory("/content/drive/MyDrive/BMG/resultados/short_order/Cu475Zr475Ag5.lammpstrj")
```

```python
traj
```

```python
frame1 = traj[0]
frame2 = traj[17]
sys1 = frame1.to_system()
sys2 = frame2.to_system()
```

```python
rmax = 10
bins = 50
val1, dist1 = sys1[0].calculate.radial_distribution_function(rmax=rmax, bins=bins)
val2, dist2 = sys2[0].calculate.radial_distribution_function(rmax=rmax, bins=bins)
```

```python
plt.plot(dist1,val1)
plt.plot(dist2,val2 + 0.05)
plt.show()
```
